# 🔍 Model Explainability using SHAP

## Objective

Machine learning models often behave like black boxes.

This notebook explains why the model predicts customer churn using SHAP values.

Business teams can understand:

- Which features increase churn risk
- Which features reduce churn risk
- Global Feature Importance
- Individual Customer Explanation

This improves trust in AI models.

```python
pip install shap
```

```python
import pandas as pd
import numpy as np

import shap
import joblib

import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("feature_dataset.csv")
```

```python
model = joblib.load("customer_churn_model.pkl")
```

```python
features = joblib.load("model_features.pkl")
```

```python
X = df[features]
```

```python
explainer = shap.TreeExplainer(model)
```

```python
shap_values = explainer.shap_values(X)
```

```python
shap.summary_plot(

    shap_values,

    X

)
```

```python
shap.summary_plot(

    shap_values,

    X,

    plot_type="bar"

)
```
